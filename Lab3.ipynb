```python
#Multinomial Distribution
from numpy import random 
#n= No. of possible outcomes 
#pvals= list of probabilities outcomes = [1/6,1/6,1/6,1/6,1/6,1/6]
x=random.multinomial (n=20,pvals=[1/6,1/6,1/6,1/6,1/6,1/6])
#size= shape of the returned array
x=random.multinomial(20,[1/6]*6,size=1)
print(x)
```

```python
#Dice landed 4 and 2 twice 
from numpy import random 
import matplotlib.pyplot as plt 
import seaborn as sns
sns .displot(random.multinomial (n=20,pvals=[1/6,1/6,1/6,1/6,1/6,1/6]))
plt.show()
```

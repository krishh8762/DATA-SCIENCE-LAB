```python
import pandas as pd 
import numpy as np 
import seaborn as sns 
import matplotlib.pyplot as plt 
#sklearn---->python Library---->for--->machine learing 
from sklearn.model_selection import train_test_split
```

```python
df=pd.read_csv(url)
print(df)
attributes=["sepal_length","sepal_width","petal_length","class"]
dataset=pd.read_csv(url,names=attributes)
dataset.column=attributes
dataset
```

```python
iris=pd.read_csv('/Users/krishkumar/Documents/iris.csv')
iris.head(5)
```

```python
#Plot---->histogram (univariate)
iris['petal_length'].hist(bins=20)
#histogram plot using matplotlib
plt.hist(data=dataset,x='petal_length',bins=20)
#Using seaborn
sns.histplot(dataset["petal_length"],bins=20)
#Histogram (UD) of each variables(columns) using pandas
iris.hist(bins=20,figsize=(10,10))
#modify
iris.groupby('species').petal_length.plot.hist(alpha=0.4)
plt.xlabel('Petal Length (cm)')
plt.suptitle("histogram of Petal length (cm) for different species")
plt.legend()
plt.show()
```

```python
#Box Plot 
sns.boxplot(data=iris,y="sepal_length")
min(iris['sepal_length'])
np.median(iris['sepal_length'])#first (Q1), Second(Q2), Third(Q3)
iris.sepal_length.quantile([0.25,0.5,0.75])
np.percentile(iris.sepal_length,[25,50,75])
#Box plot for each categories 
#use---->seaborn---->use----boxplot---->take data=iris----give me the sepal length of different species
sns.boxplot(data=iris,x='species',y='sepal_length')
plt.show
```

```python
#Violinn Plot
sns.violinplot(data=iris,x='species',y='sepal_length')
plt.show()
```

```python
#BIVARITATE RELATIONSHIP
#SCATTERPLOT
#TO SEE THE RELATIONSHIP BETWEEN TWO VARIABLE (BI- VARIATE)
sns.set_style('darkgrid')
sns.scatterplot(data=iris,x='petal_length',y='sepal_length',hue='species')
#Hue=species----->scatter plot in the graph whill be colored as per the species 
plt.show()
```

```python
#BIVERTE RELATIONSHIP 
#SCATTERPLOT 
#TO SEE THE RELATIONSHIP BETWEEN TWO VARIABLE (BI-VARIATE)
sns.set_style('darkgrid')
sns.scatterplot(data=iris,x='petal_length',y='sepal_width',hue='species')
#Hue=species---->scatter plot in the graph will be colored as per the species 
plt.show()
```

```python
#Scatter plot using matplotlib pandas and seaborn
sns.scatterplot(data=iris,x='petal_length',y='sepal_length',hue='species')
#MATPLOTLIB
plt.scatter(data=iris,x='petal_length',y='petal_width',c='green',marker='+')
#PANDAS
iris.plot.scatter(x='sepal_length',y='sepal_width',c='red',marker='+',s=15)
plt.show()
```

```python
#PAIR PLOT 
sns.pairplot(iris,hue="species",height=2,palette='colorblind')
plt.show()
```

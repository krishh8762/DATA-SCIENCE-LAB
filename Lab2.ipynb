```python
#Program to calculate the mean of an array using numpy
import numpy as np
arr1=np.array([12,234,200,100,213,10,1,112,113])
mean=np.mean(arr1)
print("the original array is ",arr1)
print("the mean of the given array is ",mean)
```

```python
#Program to calculate the median of an array using numpy
arr2=np.array([12,13,14,15,16,17,18,19,20])
med=np.median(arr2)
print("the original array is ",arr2)
print("the median value of the given array is ",med)
```

```python
#Program to calculate the mode in an array
#mode: the most repetative value in an array 
import statistics as st
arr3=np.array([10,20,30,40,50,60,10,10,100])
md=st.mode(arr3)
print("the original array is ",arr3)
print("\n")
print("the mode of the given array above is ",md)
```

```python
#Program to calculate the standdard deviation of an array 
#S.D: How far a value is deviated from the mean
#Higher the S.D: Higher is the deviation 
#Lower the S.D: Lower is the deviation
arr4=np.array([10,20,20,15,17,18,20,17,15])
Sd=np.std(arr4)
m1=np.mean(arr4)
print("the original array is ",arr4)
print("\n")
print("the mean of the given array is ",m1)
print("the calculated value of the standard deviation is ",Sd)
```

```python
#Program to calculate the percentile of an array
per=np.array([10,14,15,17,45,50,55,60,80,100,33,63,72])
#Find out the 75 percentile of the data
a=np.percentile(per,75)
print(a)
```

```python
#Task 
#take an input form the user and perform the following task
#step1: Ask the user to input size of the array
#step2: Ask the user to input values of the array
#step3: Calculate the mean, median, mode, standard deviation and percentile of the array
#step4: Find the dimensions of the array given by the user
```

```python
import numpy as np
from statistics import mode

# Ask the user to input the size of the array
size = int(input("Enter the size of the array: "))

# Ask the user to input values
user_list = []

print(f"Enter {size} elements one by one:")

for i in range(size):
    element = float(input(f"Enter {i+1}: "))
    user_list.append(element)

# Convert list to numpy array
arr1 = np.array(user_list)

# Calculate statistical values
mean = np.mean(arr1)
med = np.median(arr1)
md = mode(arr1)
sd = np.std(arr1)
per = np.percentile(arr1, 75)

# Display results
print("\nResults:")
print("The original array is:", arr1)
print("The mean of the given array is:", mean)
print("The median of the given array is:", med)
print("The mode of the given array is:", md)
print("The standard deviation of the given array is:", sd)
print("The 75 percentile of the given array is:", per)
```

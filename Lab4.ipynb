```python
#Open (filename,mode)
#mode:--->
#r:open an existing file foe read op
#w:write op
#r+:read and write (overide the existing file)
#w+: write and read (overide the ex.file)
#x: create a specific file 
#"t": text file 
#"b": Binary file 
# Open a file in write mode
f = open("samplefile.text", "w")
f.write("Hello World")
f.close()

# Open the file in read mode
f = open("samplefile.text", "r")
print(f.read())
f.close()
```

```python
#read two lines of the file
f=open("samplefile.text","r")
print(f.readline())
print(f.readline())
f.close()
```

```python
#read two lines in the file 
f=open ("samplefile.text","r")
print(f.readline())
#read line () is used to read a line 
print (f.readline())
#read the file line by line 
# Read the file line by line

f = open("samplefile.text", "r")

for x in f:
    print(x)

f.close()
```

```python
# Python program to illustrate the split() function

# Open the file and store its path
path = "samplefile.text"

# Opening the file
with open(path, "r") as file:

    # Store all lines of the file
    data = file.readlines()

    # Print each line
    for line in data:
        print(line)
       
```

```python
import pandas as pd

# Creating dataframe
df = pd.DataFrame({
    'name': ['ajay', 'vijay', 'sujal'],
    'course': ['eng', 'bio', 'bsc'],
    'location': ['delhi', 'lucknow', 'arunachal']
})

# Specifying compression method and archive name
compression_opts = dict(
    method='zip',
    archive_name='a.csv'
)

# Creating ZIP file
df.to_csv('a.zip', index=False, compression=compression_opts)
```
